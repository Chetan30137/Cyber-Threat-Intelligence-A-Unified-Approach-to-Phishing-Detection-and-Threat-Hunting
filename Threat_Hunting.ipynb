```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from torch_geometric.nn import GCNConv, global_mean_pool
from torch_geometric.data import Data, Dataset, DataLoader
from torch_geometric.utils import dense_to_sparse
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, roc_curve, auc
```

### Load dataset

```python
# Load Dataset
dataset_path = r"D:\SDP\ThreatHunting\IIoT_Network_Datasets.csv"  # Adjust for Colab if needed
df = pd.read_csv(dataset_path)

# Encode categorical values
encoder = LabelEncoder()
df['src_ip_address'] = encoder.fit_transform(df['src_ip_address'])
df['dst_ip_address'] = encoder.fit_transform(df['dst_ip_address'])
df['protocol'] = encoder.fit_transform(df['protocol'])
df['attack'] = df['attack'].apply(lambda x: 1 if x == "Attack" else 0)  # Binary classification

# Normalize numerical features
scaler = MinMaxScaler()
df[['src_port', 'dst_port', 'bytes_trans']] = scaler.fit_transform(df[['src_port', 'dst_port', 'bytes_trans']])

# Prepare Graph Data
node_features = torch.tensor(df[['src_ip_address', 'dst_ip_address', 'protocol', 'src_port', 'dst_port', 'bytes_trans']].values, dtype=torch.float)
labels = torch.tensor(df['attack'].values, dtype=torch.long)

# Split data into train and test
train_nodes, test_nodes, train_labels, test_labels = train_test_split(node_features, labels, test_size=0.2, random_state=42)

# Construct adjacency matrix (fully connected graph for simplicity)
adj_matrix = torch.ones(train_nodes.shape[0], train_nodes.shape[0])  # Dense graph assumption
edge_index, _ = dense_to_sparse(adj_matrix)  # Convert to sparse format

# Define Graph Dataset Class
class ThreatDataset(Dataset):
    def __init__(self, features, labels):
        super(ThreatDataset, self).__init__()
        self.features = features
        self.labels = labels

    def len(self):
        return len(self.labels)

    def get(self, idx):
        x = self.features[idx].unsqueeze(0)  # Make it 2D
        y = self.labels[idx]
        edge_index = torch.tensor([[0], [0]], dtype=torch.long)  # Self-loop adjacency
        return Data(x=x, edge_index=edge_index, y=y)


# Create DataLoader
train_dataset = ThreatDataset(train_nodes, train_labels)
test_dataset = ThreatDataset(test_nodes, test_labels)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Define Graph Neural Network Model
class GCNThreatHunting(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(GCNThreatHunting, self).__init__()
        self.conv1 = GCNConv(input_dim, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, hidden_dim)
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.relu = nn.ReLU()

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = self.conv1(x, edge_index)
        x = self.relu(x)
        x = self.conv2(x, edge_index)
        return self.fc(x)  # No global pooling


# Initialize model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
input_dim = train_nodes.shape[1]
model = GCNThreatHunting(input_dim, hidden_dim=64, output_dim=2).to(device)
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.CrossEntropyLoss()

# Training Loop
num_epochs = 50
for epoch in range(num_epochs):
    model.train()
    epoch_loss = 0
    for batch in train_loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        outputs = model(batch)  
        loss = criterion(outputs, batch.y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {epoch_loss:.4f}")


# Evaluation
model.eval()
all_preds, all_labels, all_probs = [], [], []
with torch.no_grad():
    for batch in test_loader:
        batch = batch.to(device)
        outputs = model(batch)
        probs = F.softmax(outputs, dim=1)[:, 1]  # Probabilities for "Attack" class
        preds = torch.argmax(outputs, dim=1)
        
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(batch.y.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())  # Store probabilities for ROC

# Compute Metrics
accuracy = accuracy_score(all_labels, all_preds)
f1 = f1_score(all_labels, all_preds)
precision = precision_score(all_labels, all_preds)
recall = recall_score(all_labels, all_preds)
print(f"Accuracy: {accuracy:.4f}, F1 Score: {f1:.4f}, Precision: {precision:.4f}, Recall: {recall:.4f}")

# Compute ROC Curve
fpr, tpr, _ = roc_curve(all_labels, all_probs)
roc_auc = auc(fpr, tpr)

# Plot ROC Curve
plt.figure(figsize=(8,6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')  # Diagonal baseline
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve for Threat Hunting Model')
plt.legend(loc="lower right")
plt.grid()
plt.show()

# Save Model
torch.save(model.state_dict(), r"D:\SDP\ThreatHunting\model\threat_hunting_gnn.pth")


```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torch
from pytorch_tabnet.tab_model import TabNetClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_curve, auc, classification_report
```

```python
# ✅ Load Dataset
df = pd.read_csv(r"D:/SDP/ThreatHunting/Threat_Hunting.csv")  # Change path if needed

# ✅ Normalize Column Names
df.columns = df.columns.str.lower().str.strip()

# ✅ Verify Columns
print("Column Names:", df.columns.tolist())

# ✅ Ensure 'label' column exists
if "label" not in df.columns:
    raise ValueError(f"Column 'label' not found! Available columns: {df.columns.tolist()}")

# ✅ Separate Features & Labels
X = df.drop(columns=["label"])
y = df["label"]

# ✅ Handle Missing Values
X.fillna(0, inplace=True)

# ✅ Convert Categorical Features
X = pd.get_dummies(X)

# ✅ Encode Labels for TabNet
le = LabelEncoder()
y = le.fit_transform(y)

# ✅ Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X.values, y, test_size=0.2, random_state=42, stratify=y)

# ✅ Standardize Features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# ✅ Initialize TabNet Model
model = TabNetClassifier(
    optimizer_fn=torch.optim.Adam,
    optimizer_params=dict(lr=0.02),
    scheduler_params={"step_size": 10, "gamma": 0.9},
    scheduler_fn=torch.optim.lr_scheduler.StepLR,
    mask_type="entmax"
)

# ✅ Train Model
model.fit(
    X_train, y_train,
    eval_set=[(X_test, y_test)],
    eval_metric=["accuracy"],
    max_epochs=50,
    patience=10,
    batch_size=1024,
    virtual_batch_size=128,
    num_workers=0
)

# ✅ Save the trained model
model_path = "D:/SDP/ThreatHunting/model/threat_hunting_model.pth"  # Change path if needed
torch.save(model, model_path)
print(f"Model saved to {model_path}")

# ✅ Make Predictions
y_prob = model.predict_proba(X_test)
y_pred = np.argmax(y_prob, axis=1)

# ✅ Compute Performance Metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average="macro")
recall = recall_score(y_test, y_pred, average="macro")
f1 = f1_score(y_test, y_pred, average="macro")

# ✅ Display Metrics
metrics_df = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1-score"],
    "Score": [accuracy, precision, recall, f1]
})
print("\nPerformance Metrics:\n", metrics_df)

# ✅ Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=le.classes_, yticklabels=le.classes_)
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix")
plt.show()

# ✅ Multiclass ROC Curve
y_test_bin = label_binarize(y_test, classes=range(len(le.classes_)))
roc_auc_scores = {}

plt.figure(figsize=(8, 6))
for i, class_label in enumerate(le.classes_):
    fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_prob[:, i])
    roc_auc = auc(fpr, tpr)
    roc_auc_scores[class_label] = roc_auc
    plt.plot(fpr, tpr, lw=2, label=f"Class {class_label} (AUC = {roc_auc:.2f})")

# ✅ Random classifier reference line
plt.plot([0, 1], [0, 1], color="gray", linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve (Multiclass)")
plt.legend(loc="lower right")
plt.show()

# ✅ Print ROC AUC Scores per Class
print("\nROC AUC Scores per Class:")
for class_label, auc_score in roc_auc_scores.items():
    print(f"Class {class_label}: {auc_score:.4f}")

# ✅ Classification Report
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred, target_names=le.classes_))

# ✅ Load the trained model (if needed)
def load_model(model_path):
    loaded_model = torch.load(model_path)
    loaded_model.eval()  # Set to evaluation mode
    print("Model loaded successfully!")
    return loaded_model

# Example Usage
# loaded_model = load_model("/content/tabnet_model.pth")
```

# Using MultiModal

```python
# Imports
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder, label_binarize
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_curve, auc, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt
import seaborn as sns
from pytorch_tabnet.tab_model import TabNetClassifier
from itertools import cycle
```

```python
# 1️⃣ Step 1: Load Dataset
df = pd.read_csv("D:/SDP/ThreatHunting/Threat_Hunting.csv")
df.columns = df.columns.str.lower().str.strip()
assert "label" in df.columns, "Label column missing!"

X_raw = df.drop(columns=["label"])
y_raw = df["label"]

# 2️⃣ Step 2: AutoEncoder for Feature Denoising
class AutoEncoder(nn.Module):
    def __init__(self, input_dim):
        super(AutoEncoder, self).__init__()
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, input_dim)
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return encoded, decoded

# Preprocessing
X = pd.get_dummies(X_raw)
X.fillna(0, inplace=True)
le = LabelEncoder()
y = le.fit_transform(y_raw)

X_train, X_test, y_train, y_test = train_test_split(X.values, y, test_size=0.2, random_state=42, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Train AutoEncoder
device = 'cuda' if torch.cuda.is_available() else 'cpu'
autoencoder = AutoEncoder(input_dim=X_train.shape[1]).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(autoencoder.parameters(), lr=0.001)
X_train_tensor = torch.FloatTensor(X_train).to(device)

for epoch in range(30):
    encoded, decoded = autoencoder(X_train_tensor)
    loss = criterion(decoded, X_train_tensor)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if (epoch+1) % 5 == 0:
        print(f"Autoencoder Epoch {epoch+1}, Loss: {loss.item():.4f}")

# Extract compressed features
with torch.no_grad():
    X_train_encoded, _ = autoencoder(X_train_tensor)
    X_test_encoded, _ = autoencoder(torch.FloatTensor(X_test).to(device))
    X_train_encoded = X_train_encoded.cpu().numpy()
    X_test_encoded = X_test_encoded.cpu().numpy()

# 3️⃣ Train TabNetClassifier
tabnet_model = TabNetClassifier(
    optimizer_fn=torch.optim.Adam,
    optimizer_params=dict(lr=0.01),
    scheduler_params={"step_size":10, "gamma":0.9},
    scheduler_fn=torch.optim.lr_scheduler.StepLR,
    mask_type='entmax'
)

tabnet_model.fit(
    X_train_encoded, y_train,
    eval_set=[(X_test_encoded, y_test)],
    eval_metric=['accuracy'],
    max_epochs=50,
    patience=10,
    batch_size=1024,
    virtual_batch_size=128
)

torch.save(tabnet_model, "D:/SDP/ThreatHunting/model/threat_hunting_multimodal.pth")

# 4️⃣ Evaluation
y_pred = tabnet_model.predict(X_test_encoded)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=le.classes_))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 7))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=le.classes_, yticklabels=le.classes_)
plt.title("Confusion Matrix")
plt.ylabel("True Label")
plt.xlabel("Predicted Label")
plt.show()

# ✅ Multiclass ROC Curve
y_test_bin = label_binarize(y_test, classes=range(len(le.classes_)))
y_score = tabnet_model.predict_proba(X_test_encoded)
n_classes = y_test_bin.shape[1]

fpr, tpr, roc_auc = {}, {}, {}
for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_score[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

plt.figure(figsize=(10, 8))
colors = cycle(['aqua', 'darkorange', 'cornflowerblue', 'green', 'red', 'purple', 'brown', 'pink', 'gray'])

for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2,
             label=f'{le.classes_[i]} (AUC = {roc_auc[i]:0.2f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlim([-0.05, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Multiclass ROC Curve')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()

# ✅ Evaluation Matrix (Accuracy, Precision, Recall, F1)
eval_matrix = {
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision (Macro)": precision_score(y_test, y_pred, average='macro'),
    "Recall (Macro)": recall_score(y_test, y_pred, average='macro'),
    "F1 Score (Macro)": f1_score(y_test, y_pred, average='macro'),
    "Precision (Micro)": precision_score(y_test, y_pred, average='micro'),
    "Recall (Micro)": recall_score(y_test, y_pred, average='micro'),
    "F1 Score (Micro)": f1_score(y_test, y_pred, average='micro')
}

eval_df = pd.DataFrame(eval_matrix, index=["Score"]).T
print("\n📊 Evaluation Matrix:")
print(eval_df)
```

```python
import pandas as pd
import torch
import torch.nn.functional as F
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from sklearn.preprocessing import LabelEncoder

# Load dataset
df = pd.read_csv("D:/SDP/ThreatHunting/Threat_Hunting.csv")

# Create formatted input string per row
def create_text_input(row):
    return (
        f"src_ip={row['orig_ip']} dst_ip={row['resp_ip']} "
        f"port={row['resp_port']} protocol={row['proto']} "
        f"service={row['service']}"
    )

df['text_input'] = df.apply(create_text_input, axis=1)

# Encode labels
le = LabelEncoder()
df['encoded_label'] = le.fit_transform(df['label'])

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=len(le.classes_))

# If you have fine-tuned model weights, uncomment below and make sure it's the correct DistilBERT checkpoint
# model.load_state_dict(torch.load("D:/SDP/ThreatHunting/model/threat_hunting_multimodal.pth", map_location=torch.device("cpu")))

model.eval()

# Prediction function using BERT
def predict_user_input(text_input):
    inputs = tokenizer(text_input, return_tensors="pt", truncation=True, padding=True)
    with torch.no_grad():
        outputs = model(**inputs)
        probs = F.softmax(outputs.logits, dim=1)
    predicted_class = torch.argmax(probs, dim=1).item()
    confidence = probs[0][predicted_class].item()
    predicted_label = le.classes_[predicted_class]
    return predicted_label, confidence

# Accept user input at runtime
src_ip = input("Enter source IP (src_ip): ")
dst_ip = input("Enter destination IP (dst_ip): ")
port = input("Enter port: ")
protocol = input("Enter protocol: ")
service = input("Enter service: ")

# Format the input
user_input = f"src_ip={src_ip} dst_ip={dst_ip} port={port} protocol={protocol} service={service}"
print("\n🧪 Testing with user-provided input:")
print(user_input)

# Predict
predicted_label, confidence = predict_user_input(user_input)
print(f"\n🔐 Predicted Threat Category: {predicted_label}")
# print(f"📈 Confidence Score: {confidence * 100:.2f}%")
```
